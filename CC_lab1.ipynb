Assignment 6.1

```python
def addodd(n):
  sum = 0
  for i in range(1,n+1):
    if(i % 2 != 0):
      sum += i
  return sum

print("addodd(10)",addodd(10))
```

Assignment 6.2

```python
def addPrime(n):
  s = 0
  for i in range(1,n+1):
    flag = 0
    for j in range(2,int(i**1/2 + 1)):
      if(i % j==0):
        flag = 1
        break
    if(flag ==0):
      s += i

  return s

print("addPrime(10)",addPrime(10))

```
